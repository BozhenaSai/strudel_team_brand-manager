### 2. Define pipeline configuration
Set the source and target catalogs, layer schemas, and the eight TPC-H tables to load.

```python
SOURCE_CATALOG = "samples"
SOURCE_SCHEMA = "tpch"

TARGET_CATALOG = "workspace"

BRONZE_SCHEMA = "brand_bronze"
SILVER_SCHEMA = "brand_silver"
GOLD_SCHEMA = "brand_gold"

TABLES = [
    "customer", "lineitem", "nation", "orders",
    "part", "partsupp", "region", "supplier",
]
```

### 3. Create target schemas
Create Bronze, Silver, and Gold schemas if they do not already exist.

```python
for schema in [BRONZE_SCHEMA, SILVER_SCHEMA, GOLD_SCHEMA]:
    spark.sql(
        f"CREATE SCHEMA IF NOT EXISTS "
        f"`{TARGET_CATALOG}`.`{schema}`"
    )
    print(f"Готово: {TARGET_CATALOG}.{schema}")
```
